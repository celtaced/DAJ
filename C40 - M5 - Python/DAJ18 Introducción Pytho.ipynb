```python
#Ejercicio
# Si se hace una venta por menos de 3 unidades el precio de cada articulo es de 4.50,
# si es hasta 144 unidades se aplicara un descuento del 3% y si se vende mas 144
# unidades, se aplicara un descuento del 5%, imprima el valor total de la venta

Un = 300
Precio = 4.5
Subtotal = Un * Precio

if Un < 3:
  descuento = 0
elif Un < 144:
  descuento = Subtotal * 0.03
else:
  descuento = Subtotal * 0.05

print("Subtotal: ", Subtotal)
print("Descuento: ", descuento)
print("Pagar: ", Subtotal - descuento)
```

```python
# Generar un programa para hacer las tablas de multiplicacion
multiplicando = int(input("Ingrese el numero de la tabla a generar"))
multiplicadores = list(range(1, 11))
for multiplicador in multiplicadores:
  print(f"{multiplicando} X {multiplicador} = {multiplicando * multiplicador}")

```

```python
# Genera un script en el que se capture la cantidad de de tablas de multiplicar a generar.
CantidadTablas = int(input("Cantidad de Tablas"))
multiplicandos = list(range(1,CantidadTablas+1))
multiplicadores = list(range(1,11))
for Mo in multiplicandos:
  print(f"Tabla del {Mo}")
  for Mr in multiplicadores:
    print(f"{Mo} X {Mr} = {Mo * Mr}")
  print()
print("Proceso terminado")
```
