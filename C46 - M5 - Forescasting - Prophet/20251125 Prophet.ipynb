```python
import pandas as pd
from prophet import Prophet
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/facebook/prophet/main/examples/example_wp_log_peyton_manning.csv')
df
```

```python
# Inicializar el modelo Prophet
# m = Prophet(interval_width=0.95)
m = Prophet()
```

```python
# Entreanar el modelo
m.fit(df)
```

```python
# Prerara un DF que recibira la informacion del forescasting
futuro = m.make_future_dataframe(periods=365, include_history=False)
futuro
```

```python
proyeccion = m.predict(futuro)
proyeccion
```

```python
# Mostrar solo las columnas relevantes
proyeccion[['ds','yhat','yhat_lower','yhat_upper']]
```

```python
# # Prerara un DF que recibira la informacion del forescasting
futuroMensual = m.make_future_dataframe(periods=12, include_history=False, freq='M')
futuroMensual
```

```python
prediccionmensual = m.predict(futuroMensual)
prediccionmensual[['ds','yhat','yhat_lower','yhat_upper']]
```
