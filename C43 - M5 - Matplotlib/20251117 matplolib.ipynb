```python
# Lectura de archivo csv
df = pd.read_csv("/kaggle/input/titanic/train.csv" )
dfTs = pd.read_csv("/kaggle/input/titanic/test.csv")
dfGs = pd.read_csv("/kaggle/input/titanic/gender_submission.csv")
```

```python
dfT1 = pd.merge(dfTs, dfGs, on="PassengerId", how="inner")
dfT2 = dfT1[["PassengerId", "Survived", "Pclass", "Name", "Sex", "Age", "SibSp", "Parch", "Ticket", "Fare", "Cabin", "Embarked"]]
dfAllData = pd.concat([df, dfT2], ignore_index=True)
dfAllData
```

```python
dfPie = dfAllData[["Pclass","Name"]].groupby("Pclass").count()
dfPie
```

```python
# visualizacion sencilla
plt.figure()
plt.pie(
    dfPie["Name"], # Valor
    labels = dfPie.index,
    autopct = "%1.1f%%",
    startangle = 90,
    colors=["red","green","bisque"]
)
```
