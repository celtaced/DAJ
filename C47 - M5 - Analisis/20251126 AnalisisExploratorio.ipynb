```python
df = pd.read_csv('melb_data.csv')
df
```

```python
# Mostrar información básica del DataFrame
df.info()
```

```python
# Creamos una lista con los nombres de las columnas que son de tipo object (Textos)
ColumnsObject = df.select_dtypes(include=['object']).columns.to_list()
ColumnsObject
```

```python
# Creamos una lista con los nombres de las columnas que son de tipo numérico
ColumnsNumber = df.select_dtypes(include=['number']).columns.to_list()
ColumnsNumber
```

```python
count_nulls = df.isnull().sum()
# count_nulls

cols_nulls = count_nulls[count_nulls > 0]
# cols_nulls

porcentaje_nulls = (cols_nulls / len(df)) * 100

porcentaje_nulls
```

```python
ax = porcentaje_nulls.plot(kind='barh', figsize=(6, 6), color='orange')
ax.set_ylabel('Porcentaje de valores nulos (%)')
ax.set_title('Porcentaje de valores nulos por columna')
ax.bar_label(ax.containers[0], fmt='%.2f%%')
```

```python
# Correlaccion de variables
x = df[ColumnsNumber]

corr = x.corr()
corr
```

```python
ax = sns.heatmap(corr, annot=True, fmt=".1f", cmap='coolwarm', cbar_kws={"shrink": .8})
```

```python
# Tratamientos a las columans

# 1. Eliminar columnas no utiles

df01 = df.drop(columns=['Address', 'SellerG', 'Date', 'Postcode', 'Suburb', 'CouncilArea', 'Regionname', 'Propertycount'])
df01
```

```python
df01.isnull().sum()
```

```python
df_sus_media = df01[['Car','BuildingArea','YearBuilt']].copy()
df_sus_media.median()
```

```python
df_sus_media
```

```python
df_sus_media['BuildingArea'] = df_sus_media['BuildingArea'].mask(df_sus_media['BuildingArea'].isnull(), df_sus_media['BuildingArea'].median())
df_sus_media
```

```python
df_sus_media['YearBuilt'] = df_sus_media['YearBuilt'].mask(df_sus_media['YearBuilt'].isnull(), df_sus_media['YearBuilt'].median())
df_sus_media
```

```python
df_sus_media.isnull().sum() 
```
