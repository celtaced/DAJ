```python
# Crear un funcion que permita, calcular el iva, calcular la venta mas iva y que considere si el cliente es excento o no. Ademas, debe validar que el monto sea mayor a 0
  # TipoTrx a> Solo Calcular IVA B> Calcular el total de venta
  # ClienteExcento True Excento False Gravado
def CalculoIVA(Monto, TipoTrx="A", ClienteExcento=False):
  if ClienteExcento:
    if TipoTrx == "A":
    # =   Asignacion
    # ==  Comparacion
      return 0
    elif TipoTrx == "B":
      return Monto
    else:
      return "Tipo de transaccion no existe"
  else:
    if TipoTrx == "A":
      return round(Monto * 0.13,2)
    elif TipoTrx == "B":
      return round(Monto * 1.13,2)
    else:
      return "Tipo de transaccion no existe"


print(CalculoIVA(100,"C",False))
```

```python
# Generar una calculadora que genere el Interes Simple haciendole llegar el
# valor principal, el plazo y la tasa de interes.
# Para el interes, si no lo envia el usuario, se utilizara una tasa del 9%,
# Si no nos envia el plazo, asumir que son 7 agnos
# I = Pin (plazo en agno)

def InteresSimple(Monto, TasaInteres=0.09, Plazo=7):
  return Monto * TasaInteres * Plazo

print(InteresSimple(float(input("prueba"))))
```

```python
# Generar un funcion que devuelva la cuatro operaciones basicas de dos numeros enviados
def OperacionesBasicas(a, b):
  Suma = a + b
  Resta = a - b
  Producto = a * b
  Division = a / b if b != 0 else "NA"
  return Suma, Resta, Producto, Division

OperacionesBasicas(2, 3)

s, r, m, d = OperacionesBasicas(4, 0)

print(s)
print(r)
print(m)
print(d)
```

```python
# No es necesario que una funcion sea un calculo, sino que puede ser una accion
def MayorEdad(Edad):
  if Edad < 18:
    print("Menor de Edad")
  else:
    print("Mayor de Edad")

Edades = [12, 11, 15, 21, 22, 18, 22, 99, 45]
for Edad in Edades:
  MayorEdad(Edad)
```

```python
# Una Funcion puede llamar a otra funcion
def Interes(P,i,n):
  return P*i*n

def TotalAPagar(P,i,n):
  I = Interes(P,i,n)
  return P + I

TotalAPagar(1000,0.05,5)
```
