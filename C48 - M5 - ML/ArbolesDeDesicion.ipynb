```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb
from sklearn.tree import plot_tree
from sklearn import preprocessing
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
from sklearn.tree import DecisionTreeRegressor,DecisionTreeClassifier
from sklearn.ensemble import RandomForestRegressor,RandomForestClassifier
from sklearn.model_selection import learning_curve
from sklearn.model_selection import ShuffleSplit
from sklearn.model_selection import cross_val_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import KFold
```

```python
df = pd.read_csv('pima-indians-diabetes.csv')
df
```

```python
df[["diabetes"]].groupby(df["diabetes"]).count()
```

```python
df.info()
```

```python
# Separar los resultados de las caractericas
x = df.drop("diabetes", axis=1).copy()
y = df["diabetes"].copy()
x
```

```python
y
```

```python
# Separando nuestro df en partes para entramiento y para pruebas
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_test
```

```python
# Crear el modelo de árbol de decisión
dfEntrenamiento = DecisionTreeClassifier(random_state=42, max_depth=14, max_features='sqrt', min_samples_split=1.0)
```

```python
dfEntrenamiento.fit(x_train, y_train)
```

```python
# Realizar pruebas de predicción
testeteo = dfEntrenamiento.predict(x_test)
testeteo
```

```python
y_test
```

```python
reporte_clasificacion = classification_report(y_test, testeteo)
print(reporte_clasificacion)
```
